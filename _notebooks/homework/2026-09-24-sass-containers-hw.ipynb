# SASS: Containers

Original practice built around the same lesson objectives, using my own examples and wording.

## Popcorn Hack: Meet-Day Card

Use the OCS container/card structure to group related information instead of leaving loose text on the page.

```html
<div class="ocs__container">
  <section class="ocs__card">
    <h2 class="ocs__section-title">Cross Country Meet Day</h2>
    <p class="ocs__description">Bus leaves at 11:15 AM. Bring uniform, spikes, water, and a snack.</p>
  </section>
</div>
```

## Homework Hack: Student Event Hub

This version groups three pieces of information into one organized container: schedule, reminders, and contact information.

```html
<div class="ocs__container">
  <div class="ocs__grid ocs__grid--standard">
    <div class="ocs__grid-cell ocs__grid-cell--header">Senior Event Hub</div>
    <div class="ocs__grid-cell"><strong>Schedule</strong><br>Check-in 4:30 PM<br>Start 5:00 PM</div>
    <div class="ocs__grid-cell ocs__grid-cell--accent"><strong>Bring</strong><br>ID, water, charger</div>
    <div class="ocs__grid-cell ocs__grid-cell--muted"><strong>Questions</strong><br>Ask an event leader before check-in.</div>
  </div>
</div>
```

**Why containers help:** related content stays visually grouped, making the page easier to scan.
