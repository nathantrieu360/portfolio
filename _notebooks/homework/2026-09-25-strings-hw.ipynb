## Homework Hack: Decode a Final Record

Parse a CSV-style string, convert the numeric field, and rebuild it as a readable sentence.

```python
record = "XC,5K,17.08,varsity"
parts = record.split(",")
team = parts[0]
event = parts[1]
time_minutes = float(parts[2])
level = parts[3].title()
print(f"{team} {event}: {time_minutes:.2f} minutes, {level} level.")
```
