# SASS: Buttons

Original practice using my own examples and wording.

## Popcorn Hack 1: Base Button

```html
<a class="ocs__btn" href="#projects">View Projects</a>
```

## Popcorn Hack 2: Button Group

```html
<div class="ocs__links">
  <a class="ocs__btn" href="#about">About</a>
  <a class="ocs__btn" href="#contact">Contact</a>
</div>
```

## Popcorn Hack 3: Wide Action Bar

```html
<div class="ocs__links ocs__links--wide">
  <a class="ocs__btn" href="#music">Music Picks</a>
  <a class="ocs__btn" href="#chess">Chess</a>
  <a class="ocs__btn" href="#running">Running</a>
</div>
```

## Homework Hack 1: Adaptive Resource Toolbar

```html
<div class="ocs__container">
  <div class="ocs__links ocs__links--wide">
    <a class="ocs__btn" href="/notes/">Class Notes</a>
    <a class="ocs__btn" href="/homework/">Homework</a>
    <a class="ocs__btn" href="/projects/">Projects</a>
  </div>
</div>
```

## Homework Hack 2: Primary vs. Secondary Actions

```html
<div class="ocs__links">
  <button class="ocs__btn">Save Changes</button>
  <button class="ocs__btn">Preview</button>
  <button class="ocs__btn">Cancel</button>
</div>
```

The main action comes first in reading order, followed by supporting actions.
