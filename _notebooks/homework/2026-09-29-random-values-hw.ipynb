## Popcorn Hack 1: Random Bib Number

```python
import random
bib = random.randint(100, 999)
print("Random bib:", bib)
```

## Popcorn Hack 2: Random Structured Workout

```python
workouts = [
    {"type":"tempo","minutes":20},
    {"type":"intervals","reps":5},
    {"type":"recovery","miles":4}
]
choice = random.choice(workouts)
print("Selected workout:", choice)
```

## Popcorn Hack 3: Random Test Action

```python
actions = ["create", "read", "update", "delete"]
print("QA action:", random.choice(actions))
```

## Homework Hack: Training App QA Simulator

```python
import random
records = [
    {"id":101,"runner":"Ava","status":"ready"},
    {"id":102,"runner":"Mason","status":"recovery"},
    {"id":103,"runner":"Leo","status":"ready"}
]
actions = ["read","update","create","delete"]
for test in range(5):
    record = random.choice(records)
    action = random.choice(actions)
    print(f"Test {test+1}: {action.upper()} record {record['id']} ({record['status']})")
```

## Randomness Explanation

Repeated values are normal because each selection is independent. Running the simulator multiple times increases coverage, but random testing alone cannot guarantee every possible combination appears.
