# SASS: Toggles

Original practice using my own data and examples.

## Popcorn Hack: Fill in the Toggle Structure

A toggle needs a checkbox input, a visible track, and a readable label.

```html
<label class="ocs__toggle">
  <input class="ocs__toggle-input" type="checkbox" checked>
  <span class="ocs__toggle-track"></span>
  <span class="ocs__toggle-label">Show race splits</span>
</label>
```

## Homework Hack: Portfolio Settings Panel

```html
<div class="ocs__container">
  <h2 class="ocs__section-title">Portfolio Settings</h2>

  <label class="ocs__toggle ocs__toggle--green">
    <input class="ocs__toggle-input" type="checkbox" checked>
    <span class="ocs__toggle-track"></span>
    <span class="ocs__toggle-label">Enable study reminders</span>
  </label>

  <label class="ocs__toggle">
    <input class="ocs__toggle-input" type="checkbox">
    <span class="ocs__toggle-track"></span>
    <span class="ocs__toggle-label">Compact project cards</span>
  </label>
</div>
```

The labels explain what each control changes, and the checkbox state stores whether the setting is on or off.
