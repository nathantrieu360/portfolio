## Homework Hack: Event Registration Validator

```python
existing_codes = {"RUN1", "CHS2"}
allowed_events = {"run", "chess", "coding"}

def validate(record):
    name_ok = isinstance(record.get("name"), str) and len(record["name"].strip()) >= 2
    age_ok = isinstance(record.get("age"), int) and 10 <= record["age"] <= 18
    event_ok = record.get("event") in allowed_events
    code = record.get("code", "")
    code_ok = len(code) == 4 and code not in existing_codes
    return name_ok and age_ok and event_ok and code_ok

tests = [
 {"name":"Maya","age":15,"event":"coding","code":"COD3"},
 {"name":"J","age":15,"event":"coding","code":"COD4"},
 {"name":"Leo","age":19,"event":"run","code":"RUN1"}
]
for item in tests:
    print(item, "=>", validate(item))
```

## Boolean Logic Explanation

Breaking the rule into named Boolean checks makes it easier to see exactly why a record passes or fails.
