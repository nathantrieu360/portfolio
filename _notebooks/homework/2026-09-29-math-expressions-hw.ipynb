# Python 3.03: Math Expressions

Original practice using my own examples and wording.

## Popcorn Hack 1: Basic Arithmetic Operators

```python
laps = 12
lap_distance = 0.25
total_miles = laps * lap_distance
remaining = 5 - total_miles
print("Miles completed:", total_miles)
print("Miles remaining:", remaining)
```

## Popcorn Hack 2: Race Pace Math

```python
distance_miles = 3.1
time_minutes = 17 + 8/60
pace = time_minutes / distance_miles
print("Average pace:", round(pace, 2), "minutes per mile")
```

## Homework Hack: Event Budget Report

```python
entry_fee = 22.50
participants = 48
permit = 180
medals = 96
music = 40
revenue = entry_fee * participants
costs = permit + medals + music
remaining = revenue - costs
cost_per_runner = costs / participants
print("Revenue: $", round(revenue,2))
print("Costs: $", round(costs,2))
print("Remaining: $", round(remaining,2))
print("Cost per runner: $", round(cost_per_runner,2))
```

## Homework Results

The expressions use multiplication for total revenue, addition for combined costs, subtraction for remaining money, and division for per-runner cost.
