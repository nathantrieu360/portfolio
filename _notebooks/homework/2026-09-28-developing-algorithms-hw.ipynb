## Popcorn Hack 1: Translate Pseudocode

```python
scores = [8, 6, 9, 7]
total = 0
for score in scores:
    total += score
print("Total:", total)
```

## Popcorn Hack 2: Predict, Then Fix

```python
attendance = [18, 22, 19]
count = 0
for people in attendance:
    if people >= 20:
        count += 1
print("Meetings with 20+ members:", count)
```

## Popcorn Hack 3: Counter to Collector

```python
ratings = [6, 9, 8, 5, 10]
strong = []
for rating in ratings:
    if rating >= 8:
        strong.append(rating)
print(strong)
```

## Homework Hack: Club Supply Restock Report

```python
supplies = {
    "markers": 3,
    "score_sheets": 14,
    "water_bottles": 5,
    "name_tags": 20
}
minimum = 8
restock = []
for item, amount in supplies.items():
    if amount < minimum:
        restock.append((item, minimum - amount))
print("Restock list:")
for item, needed in restock:
    print("-", item, "add", needed)
```

## Expected Homework Output

Only items below the minimum should be listed, and each line should show how many more are needed.
