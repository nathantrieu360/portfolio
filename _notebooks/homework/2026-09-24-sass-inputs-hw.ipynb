# SASS: Inputs

Original practice built around the same lesson objectives, using my own examples and wording.

## Popcorn Hack

Create one labeled input using the shared OCS input class.

```html
<label for="artist">Artist name</label>
<input id="artist" class="ocs__input" type="text" placeholder="Enter an artist">
```

## Homework Hack: Music Recommendation Form

```html
<div class="ocs__container">
  <label for="song">Song</label>
  <input id="song" class="ocs__input" type="text" placeholder="Song title">

  <label for="rating">Rating (1-10)</label>
  <input id="rating" class="ocs__input" type="number" min="1" max="10" value="8">

  <label for="why">Why recommend it?</label>
  <input id="why" class="ocs__input" type="text" placeholder="One short reason">
</div>
```

I used labels for accessibility and different input types so the browser can validate the rating range.
