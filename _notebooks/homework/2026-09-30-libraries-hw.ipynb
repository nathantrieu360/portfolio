## Homework Hack: Cross Country Season Stats Report

```python
import json
import statistics
from datetime import date
import math

def season_report(races):
    """Return a summary for race records.

    Args:
        races: list of dictionaries with course, time_seconds, and place.
    Returns:
        dictionary containing count, average time, best race, average place,
        and the date the report was generated.
    """
    times = [r["time_seconds"] for r in races]
    places = [r["place"] for r in races]
    best = min(races, key=lambda r: r["time_seconds"])
    return {
        "races": len(races),
        "average_time_seconds": round(statistics.mean(times), 1),
        "best_course": best["course"],
        "best_time_seconds": best["time_seconds"],
        "average_place": round(statistics.mean(places), 1),
        "report_date": date.today().isoformat(),
        "pace_factor": round(math.sqrt(len(races)), 2)
    }

races = [
    {"course":"Flat 5K","time_seconds":1028,"place":12},
    {"course":"Invite 5K","time_seconds":1045,"place":18},
    {"course":"League 5K","time_seconds":1019,"place":9}
]
print(json.dumps(season_report(races), indent=2))
```
