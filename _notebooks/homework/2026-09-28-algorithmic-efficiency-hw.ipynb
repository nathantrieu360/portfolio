## Homework Hack: Choose a Search Algorithm

```python
def linear_search(values, target):
    checks = 0
    for value in values:
        checks += 1
        if value == target:
            return True, checks
    return False, checks

def binary_search(values, target):
    low, high, checks = 0, len(values)-1, 0
    while low <= high:
        checks += 1
        mid = (low + high) // 2
        if values[mid] == target:
            return True, checks
        if values[mid] < target:
            low = mid + 1
        else:
            high = mid - 1
    return False, checks

bibs = list(range(100, 500))
print("Linear:", linear_search(bibs, 487))
print("Binary:", binary_search(bibs, 487))
```

## Homework Recommendation

For a sorted list of hundreds or thousands of bib numbers, binary search usually needs far fewer comparisons. Linear search is still reasonable for a short unsorted list because it requires no sorting setup.
