# SASS: Typography

Semantic HTML practice using OCS typography classes without custom CSS.

## Popcorn Hack

Refactor the content using semantic HTML and OCS typography classes.

```python
%%html
<!-- UI_RUNNER: Typography Popcorn Hack | show_code: true -->
<div class="ocs_card">
  <h2 class="ocs_section-title">Why Cats Purr</h2>
  <h3 class="ocs_lead">A Sign of Happiness</h3>
  <p class="ocs_text">Cats usually purr when they feel safe, cozy, and ready for a nap.</p>
</div>
```

## Homework Hack

Final semantic refactor using headings, paragraphs, a real ordered list, semantic emphasis, and OCS classes.

```python
%%html
<!-- UI_RUNNER: Typography Homework Hack | show_code: true -->
<div class="ocs_card">
  <h2 class="ocs_section-title">How to Pat a Cat</h2>
  <h3 class="ocs_lead">Have you ever wanted to...</h3>
  <p class="ocs_text">Pat a cat? Good for you! Here are step-by-step instructions on how to pat a cat.</p>
  <p class="ocs_text"><strong>How to pat a cat:</strong></p>
  <ol>
    <li>Put out your hand.</li>
    <li>Gently touch the cat.</li>
    <li>Move your hand back and forth.</li>
  </ol>
  <h3 class="ocs_section-title">Congrats! You pat a cat :D</h3>
</div>
```
