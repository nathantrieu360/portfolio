# SASS: Typography

Original practice built around the same lesson objectives, using my own examples and wording.

## Popcorn Hack

Use the provided typography classes to create a clear heading, lead sentence, and body copy.

```html
<section class="ocs_card">
  <h2 class="ocs_section-title">Nested Conditionals Lesson</h2>
  <p class="ocs_lead">A student-built lesson with interactive code practice.</p>
  <p class="ocs_text">The page explains nested decisions, then gives progressively harder practice problems.</p>
</section>
```

## Homework Hack: Project Spotlight

```html
<article class="ocs_card">
  <h2 class="ocs_section-title">Chess Tutoring Project</h2>
  <p class="ocs_lead">Free lessons for younger players focused on fundamentals and confidence.</p>
  <p class="ocs_description">Students work through openings, tactics, and endgames, then apply them in short practice games.</p>
  <p class="ocs_text">The hierarchy makes the title easiest to notice, the lead sentence gives the summary, and the final paragraph adds detail.</p>
</article>
```
