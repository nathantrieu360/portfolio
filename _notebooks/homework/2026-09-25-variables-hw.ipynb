## MCQ

An assignment stores a value in a variable. If an input variable changes, a previously calculated result does not automatically recalculate.

## Homework Hack: Build a Meet Point Counter

```python
team_points = 18
individual_points = 7
bonus_points = 3
total_points = team_points + individual_points + bonus_points
print("Starting total:", total_points)

bonus_points = 5
total_points = team_points + individual_points + bonus_points
print("Updated total:", total_points)
```

## Tests

```python
for bonus in [0, 2, 5]:
    total = team_points + individual_points + bonus
    print("Bonus", bonus, "=>", total)
```

## Design Thinking

I separated the component scores so each can be changed independently. The total is recalculated after an update because Python variables do not automatically keep formulas linked like a spreadsheet.
