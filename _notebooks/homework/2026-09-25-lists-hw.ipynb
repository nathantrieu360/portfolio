# Python 3.10: Lists

Original practice using my own data and examples.

## Popcorn Hack 1: Predict, Run, Fix

Indexes start at zero, and a valid index must stay inside the list.

```python
mile_splits = [328, 334, 331]
print("First mile:", mile_splits[0])
print("Last mile:", mile_splits[-1])
```

## Popcorn Hack 2: Sum and Average

```python
total = 0
for split in mile_splits:
    total += split
print("Average split:", total / len(mile_splits))
```

## Popcorn Hack 3: Count Then Filter

```python
race_times = [1050, 1018, 1035, 995, 1072]
sub_17 = 0
fast_times = []
for t in race_times:
    if t < 1020:
        sub_17 += 1
        fast_times.append(t)
print("Sub-17 results:", sub_17)
print("Times:", fast_times)
```

## Homework Hack: Indexing, Accumulator, Count, and Filter

```python
workout = [252, 254, 254, 256, 249]
print("Opening rep:", workout[0])
accumulator = 0
under_255 = 0
filtered = []
for rep in workout:
    accumulator += rep
    if rep < 255:
        under_255 += 1
        filtered.append(rep)
print("Total seconds:", accumulator)
print("Under 4:15:", under_255)
print("Filtered reps:", filtered)
```
