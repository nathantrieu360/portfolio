# Python 3.02: Data Abstractions

Original practice using my own data and examples.

## Popcorn Hack 1: Find an Item

Use an index to access one value from a collection.

```python
ratings = [1512, 1538, 1561, 1580]
print("Second recorded rating:", ratings[1])
```

## Popcorn Hack 2: Extend the Sequence

Add a new value without replacing the existing list.

```python
ratings.append(1602)
print("Updated ratings:", ratings)
```

## MCQ

A list is useful because one variable can represent an ordered collection of related values.

## Homework Hack: Grow the Rating Collection

Add a tournament rating, update one earlier value, then report the number of entries and total rating points.

```python
ratings = [1512, 1538, 1561]
ratings.append(1602)
ratings[1] = 1544
count = len(ratings)
total = sum(ratings)
print("Ratings:", ratings)
print("Entries:", count)
print("Total:", total)
```

## Tests

The main case should contain four ratings. An empty list should safely report zero entries and a total of zero.

```python
empty_ratings = []
print(len(empty_ratings), sum(empty_ratings))
```

## Design Thinking

A list keeps the data together and makes later operations such as averages, trends, and searches easier. I also tested the empty-list edge case so the logic does not depend on data always being present.
