# Python 3.07: Nested Conditionals

Original practice using my own examples and wording.

## Popcorn Hack 1: After-School Plan

```python
is_weekday = True
homework_finished = False
if is_weekday:
    if homework_finished:
        print("Free time before practice")
    else:
        print("Finish homework first")
else:
    print("Weekend plan")
```

## Popcorn Hack 2: Student Account Check

```python
account_active = True
password_correct = True
if account_active:
    if password_correct:
        print("Access granted")
    else:
        print("Reset password")
else:
    print("Account disabled")
```

## Popcorn Hack 3: Lunch and Practice

```python
lunch_packed = False
practice_today = True
if lunch_packed:
    if practice_today:
        print("Pack an extra snack for practice")
    else:
        print("Lunch is ready")
else:
    if practice_today:
        print("Buy lunch and save a snack for later")
    else:
        print("Buy lunch")
```

## Homework Hack: High School Daily Routine Dashboard

```python
def daily_plan(logged_in, homework_done, lunch_packed, practice_today):
    if not logged_in:
        return "Log in to view today's plan."
    if homework_done:
        school_task = "Homework complete"
    else:
        school_task = "Finish homework before free time"
    if lunch_packed:
        if practice_today:
            food_task = "Bring lunch plus a practice snack"
        else:
            food_task = "Bring packed lunch"
    else:
        if practice_today:
            food_task = "Buy lunch and get a practice snack"
        else:
            food_task = "Buy lunch"
    return school_task + " | " + food_task

cases = [
    (False, False, False, True),
    (True, False, True, True),
    (True, True, False, False)
]
for case in cases:
    print(daily_plan(*case))
```

## Explanation

The outer login check protects the personalized plan. Inside the logged-in branch, later decisions depend on homework, lunch, and practice status, so nested conditionals represent the dependency between choices.
