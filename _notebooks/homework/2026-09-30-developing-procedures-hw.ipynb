## Practice 1: Rename Procedures Clearly

```python
def calculate_match_points(wins, draws):
    return wins * 1 + draws * 0.5
print(calculate_match_points(4, 2))
```

## Practice 3: Split a Long Procedure

```python
def validate_rating(rating):
    return 100 <= rating <= 3000

def rating_band(rating):
    if rating >= 1800:
        return "Advanced"
    if rating >= 1400:
        return "Intermediate"
    return "Developing"

def player_summary(name, rating):
    if not validate_rating(rating):
        return "Invalid rating"
    return f"{name}: {rating} ({rating_band(rating)})"

print(player_summary("Nathan", 1600))
```

## Mini-Game Practice: Procedure Naming

```python
def gain_rating(current, change):
    return current + change

def lose_rating(current, change):
    return max(0, current - change)

rating = 1600
rating = gain_rating(rating, 18)
rating = lose_rating(rating, 7)
print("Final rating:", rating)
```

Breaking one large task into small helpers makes each piece easier to understand, test, and reuse.
