## Homework Hack: Training Intensity Planner

```python
def training_plan(soreness, sleep_hours, workout_day):
    risk = 0
    if soreness >= 7:
        risk += 4
    elif soreness >= 4:
        risk += 2
    if sleep_hours < 6:
        risk += 3
    elif sleep_hours < 7.5:
        risk += 1
    if workout_day:
        risk += 1

    if risk >= 7:
        return risk, "Recovery or coach check"
    elif risk >= 4:
        return risk, "Reduce intensity"
    elif risk >= 2:
        return risk, "Train with caution"
    else:
        return risk, "Normal plan"

for case in [(8,5.5,True),(5,8,True),(2,8,False),(4,6.5,False)]:
    print(case, "=>", training_plan(*case))
```

## Reachability Proof

The tests hit multiple branches: high risk, reduced intensity, cautious training, and normal training. Testing thresholds from highest to lowest prevents a broad condition from swallowing a more specific one.
