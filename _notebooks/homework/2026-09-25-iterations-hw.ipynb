# Python 3.08: Iterations

Original practice using my own data and examples.

## Popcorn Hack: Split Scanner

Use a loop, `continue`, a counter, and a running total.

```python
splits = [74, 76, None, 73, 79]
valid_count = 0
total = 0
for split in splits:
    if split is None:
        continue
    valid_count += 1
    total += split
print("Valid splits:", valid_count)
print("Average:", total / valid_count)
```

## Quick Check

`continue` skips the rest of the current iteration. `break` exits the loop entirely.

## Homework Hack: Workout Load Monitor

Stop scanning once a split is above the safety threshold, while tracking completed reps.

```python
rep_times = [72, 74, 75, 82, 73]
threshold = 80
completed = 0
running_total = 0
for seconds in rep_times:
    if seconds > threshold:
        print("Stop: recovery check needed at", seconds, "seconds")
        break
    completed += 1
    running_total += seconds
print("Completed reps:", completed)
if completed:
    print("Average completed split:", round(running_total / completed, 1))
```
