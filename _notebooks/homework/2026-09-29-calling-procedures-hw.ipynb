## Quick Knowledge Check

Defining a procedure creates it; calling the procedure actually runs its instructions.

## Popcorn Hack 1: Parameters and Arguments

```python
def pace_label(seconds):
    return f"{seconds // 60}:{seconds % 60:02d} per mile"
print(pace_label(329))
```

## Popcorn Hack 2: Decomposition With Helpers

```python
def normalize_name(name):
    return name.strip().title()

def member_card(name, role):
    clean_name = normalize_name(name)
    return f"{clean_name} — {role.title()}"
print(member_card("  nathan trieu  ", "president"))
```

## Homework Hack: Meet-Day Dispatcher

```python
def convert_minutes(minutes):
    return int(minutes * 60)

def arrival_time(start_minutes, travel_minutes, buffer_minutes):
    return start_minutes - travel_minutes - buffer_minutes

def format_clock(total_minutes):
    hour = (total_minutes // 60) % 24
    minute = total_minutes % 60
    suffix = "AM" if hour < 12 else "PM"
    shown = hour % 12 or 12
    return f"{shown}:{minute:02d} {suffix}"

def dispatch_message(race_time, travel, buffer):
    leave = arrival_time(race_time, travel, buffer)
    return "Leave by " + format_clock(leave)

race_start = 13 * 60 + 10
print(dispatch_message(race_start, 35, 45))
```

## Explanation

Each helper does one job, so the main dispatcher is easier to read and test. Parameters let the same procedures work for different race times and travel plans.
