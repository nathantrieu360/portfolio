# SASS: Grids

Original practice built around the same lesson objectives, using my own examples and wording.

## Popcorn Hack 1: Refactor Into a Grid

The same information is easier to compare when each value has a consistent cell.

```html
<div class="ocs__grid ocs__grid--standard">
  <div class="ocs__grid-cell ocs__grid-cell--header">Mile</div>
  <div class="ocs__grid-cell ocs__grid-cell--header">Split</div>
  <div class="ocs__grid-cell">1</div><div class="ocs__grid-cell">5:28</div>
  <div class="ocs__grid-cell">2</div><div class="ocs__grid-cell">5:34</div>
  <div class="ocs__grid-cell">3</div><div class="ocs__grid-cell">5:31</div>
</div>
```

## Popcorn Hack 2: Emphasize One Cell

Use an accent cell for the target split instead of changing the whole layout.

```html
<div class="ocs__grid ocs__grid--standard">
  <div class="ocs__grid-cell">Current pace</div>
  <div class="ocs__grid-cell ocs__grid-cell--accent">Target: 5:25</div>
</div>
```

## MCQ Check

1. A grid is useful when information needs consistent rows/columns.  
2. A modifier class changes a specific grid variation without rewriting the base component.

## Homework Hack: Pace Calculator Layout

```html
<div class="ocs__grid ocs__grid--calculator">
  <div class="ocs__grid-cell ocs__grid-cell--header">Pace Calculator</div>
  <div class="ocs__grid-cell">Distance: 3.1 miles</div>
  <div class="ocs__grid-cell">Goal time: 17:00</div>
  <div class="ocs__grid-cell ocs__grid-cell--accent">Goal pace: about 5:29/mi</div>
</div>
```

The calculator-style grid separates inputs from the result and keeps the most important number visually distinct.
